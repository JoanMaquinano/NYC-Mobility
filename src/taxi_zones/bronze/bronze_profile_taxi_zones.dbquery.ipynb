```sql
-- Row count and basic shape
SELECT COUNT(*) AS total_rows
FROM `nyc-mobility`.nyc_bronze.taxi_zones;

-- Null counts per column
SELECT
  COUNT(*) AS total_rows,
  COUNT(*) - COUNT(LocationID) AS null_location_id,
  COUNT(*) - COUNT(Borough) AS null_borough,
  COUNT(*) - COUNT(Zone) AS null_zone,
  COUNT(*) - COUNT(service_zone) AS null_service_zone
FROM `nyc-mobility`.nyc_bronze.taxi_zones;

-- Cardinality / uniqueness check (is LocationID actually a valid key?)
SELECT
  COUNT(*) AS total_rows,
  COUNT(DISTINCT LocationID) AS distinct_location_ids
FROM `nyc-mobility`.nyc_bronze.taxi_zones;

-- Duplicate LocationIDs, if any (to inspect the actual offending rows)
SELECT LocationID, COUNT(*) AS occurrences
FROM `nyc-mobility`.nyc_bronze.taxi_zones
GROUP BY LocationID
HAVING COUNT(*) > 1;

-- Value distribution for categorical columns (catches typos, unexpected values)
SELECT Borough, COUNT(*) AS row_count
FROM `nyc-mobility`.nyc_bronze.taxi_zones
GROUP BY Borough
ORDER BY row_count DESC;

SELECT service_zone, COUNT(*) AS row_count
FROM `nyc-mobility`.nyc_bronze.taxi_zones
GROUP BY service_zone
ORDER BY row_count DESC;

-- Sample of raw values to eyeball formatting issues (extra whitespace, casing, etc.)
SELECT DISTINCT Zone
FROM `nyc-mobility`.nyc_bronze.taxi_zones
ORDER BY Zone
LIMIT 20;

-- Fails the query (returns a row) if duplicates exist — useful as a gate step
SELECT 'DUPLICATE LOCATION_ID DETECTED' AS alert
WHERE (
  SELECT COUNT(*)
  FROM `nyc-mobility`.nyc_bronze.taxi_zones
) != (
  SELECT COUNT(DISTINCT LocationID)
  FROM `nyc-mobility`.nyc_bronze.taxi_zones
);
```
